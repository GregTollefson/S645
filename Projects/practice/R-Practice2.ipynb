```r
# Number of simulated observations
n <- 1000

# Generate X1, ..., Xn from N(0,1)
x <- rnorm(n, mean = 0, sd = 1)
```

```r
x <= 1
```

```r
cumsum(x <= 1)
```

```r
relative_frequency <- cumsum(x <= 1) / (1:n)
```

```r
plot(
    1:n,
    relative_frequency,
    type = "l",
    xlab = "Number of observations (n)",
    ylab = "Relative frequency",
    main = "Relative Frequency of X <= 1"
)

abline(h = 0.8413, col = "red", lty = 2)
```

Design an experiment using any continuous random variable available in R which will show how the (probability) histogram changes and gets closer to the PDF as the sample size increases

```r
# Sample sizes
sample_sizes <- c(25, 100, 1000, 10000)

# Set up a 2 x 2 plotting window
par(mfrow = c(2, 2))

for (n in sample_sizes) {

    # Generate random observations
    x <- rgamma(n, shape = 2, rate = 1)

    # Probability histogram
    hist(
        x,
        probability = TRUE,
        breaks = 20,
        main = paste("n =", n),
        xlab = "x",
        xlim = c(0, 10)
    )

    # Add the theoretical Gamma PDF
    curve(
        dgamma(x, shape = 2, rate = 1),
        add = TRUE,
        col = "red",
        lwd = 2
    )
}
```

Illustrate the converge of ECDF to true CDF for your choice of a
continuous distribution given in R

```r
sample_sizes <- c(25, 100, 1000, 10000)

par(mfrow = c(2, 2))

for (n in sample_sizes) {

    # Generate a random sample
    x <- rgamma(n, shape = 2, rate = 1)

    # Plot the empirical CDF
    plot(
        ecdf(x),
        main = paste("n =", n),
        xlab = "x",
        ylab = "CDF",
        xlim = c(0, 10),
        verticals = TRUE,
        do.points = FALSE
    )

    # Add the true Gamma CDF
    curve(
        pgamma(x, shape = 2, rate = 1),
        add = TRUE,
        col = "red",
        lwd = 2
    )
}
```

1:Write R code that computes the sample quantiles and plots the empirical
quantile function for a sample of size 100 from the standard normal distribution.
2. Then, compute the 35th and 55th empirical quantiles from your data.
3. Find the true 35th and 55th quantiles of the standard normal distribution. Are
they close to the empirical ones?
4. What would you do to get empirical quantiles closer to the theoretical ones?

```r
# Generate a sample of size 100 from N(0,1)
set.seed(123)
x <- rnorm(100)

# Plot the empirical quantile function
p <- seq(0, 1, length.out = 100)

q_emp <- quantile(x, probs = p)

plot(
    p,
    q_emp,
    type = "l",
    xlab = "Probability p",
    ylab = "Empirical Quantile",
    main = "Empirical Quantile Function"
)
```

```r
quantile(x, probs = c(0.35, 0.55))
```

```r
qnorm(c(0.35, 0.55))
```

```r
empirical <- quantile(x, probs = c(0.35, 0.55))
theoretical <- qnorm(c(0.35, 0.55))

cbind(empirical, theoretical)
```
