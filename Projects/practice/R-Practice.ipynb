```r
# Generate 100 observations from a t distribution with 4 degrees of freedom
x <- rt(100, df = 4)

# Look at the first few observations
head(x)
```

```r
library(MASS)

truehist(x)
```

```r
truehist(
    x,
    nbins = 10,
    main = "Random Sample from t(4)",
    xlab = "x"
)
```

```r
truehist(
    x,
    nbins = 10,
    main = "Random Sample from t(4)",
    xlab = "x"
)

curve(
    dt(x, df = 4),
    add = TRUE,
    lwd = 2
)
```

```r
?rt
```

```r
library(MASS)

# Probability histogram
truehist(x)

# Add theoretical t distribution with df = 4
curve(dt(x, df = 4), add = TRUE)
```

```r
library(MASS)

# Histogram of the observed sample
truehist(x)

# Theoretical t(4) density
curve(dt(x, df = 4), add = TRUE)

# Density estimated from the sample
lines(density(x), col = 2)
```
